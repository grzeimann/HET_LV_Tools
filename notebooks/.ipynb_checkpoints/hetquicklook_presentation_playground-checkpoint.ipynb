# HET quick-look guide for observers

This notebook gives a short path for checking VIRUS and LRS-2 data during the
night. Run the cells from top to bottom. You only need to set the data paths
once, choose a row from the table, and run the matching quick-look cell.

The examples use real exposure rows from VIRUS night `20260609` and LRS-2
night `20260512`. The row numbers are examples for those dates; use the table
shown in your own night to choose a different exposure.


## 1. Configure the session

Run this cell once at the start of the notebook. Change the two paths if your
raw data are stored somewhere else. The same `ql` object is used for both
instruments.


```python
from hetquicklook import QuicklookSite

ql = QuicklookSite(
    raw_roots={
        'virus': '~/data/VIRUS',
        'lrs2': '~/data/LRS2',
    },
)
```

## 2. VIRUS night

Create the night object and run the cell. It displays a table with one row per
exposure. Row numbers start at zero.


```python
VIRUSNight = ql.night('20260609', instrument='virus')
VIRUSNight
```

### Example VIRUS rows

For the example night above:

- row `5` is the WD1327-083 standard star;
- row `24` is an LDLS flat;
- row `11` is a long science exposure of IDCS J1426+3508.

Each example below selects one row, runs the automatic quick-look workflow,
and displays the image with the `coolwarm` color map.


### VIRUS standard star


```python
VIRUS_standard_exposure = VIRUSNight[5]
VIRUS_standard_product = VIRUS_standard_exposure.quicklook()
VIRUS_standard_figure = VIRUS_standard_product.plot(cmap='coolwarm')
```

### VIRUS flat


```python
VIRUS_flat_exposure = VIRUSNight[24]
VIRUS_flat_product = VIRUS_flat_exposure.quicklook()
VIRUS_flat_figure = VIRUS_flat_product.plot(cmap='coolwarm')
```

### VIRUS long science exposure


```python
VIRUS_science_exposure = VIRUSNight[11]
VIRUS_science_product = VIRUS_science_exposure.quicklook()
VIRUS_science_figure = VIRUS_science_product.plot(cmap='coolwarm')
```

### Refresh the VIRUS table later in the night

When new archives arrive, run this cell. It refreshes `VIRUSNight` in place,
so the next table and row selections include the new exposures.


```python
VIRUSNight.update()
```

## 3. LRS-2 night

The LRS-2 workflow is the same: create a night object, choose a row from its
table, and run a quick look.


```python
LRS2Night = ql.night('20260512', instrument='lrs2')
LRS2Night
```

### Example LRS-2 rows

For the example night above:

- row `25` is the HD180609 standard star;
- row `37` is an LDLS flat;
- row `27` is a long science exposure of ICRF J2039+2152.

The row numbers are zero-based and refer to the LRS-2 table above.


### LRS-2 standard star


```python
LRS2_standard_exposure = LRS2Night[25]
LRS2_standard_product = LRS2_standard_exposure.quicklook()
LRS2_standard_figure = LRS2_standard_product.plot(cmap='coolwarm')
```

### LRS-2 flat


```python
LRS2_flat_exposure = LRS2Night[37]
LRS2_flat_product = LRS2_flat_exposure.quicklook()
LRS2_flat_figure = LRS2_flat_product.plot(cmap='coolwarm')
```

### LRS-2 long science exposure


```python
LRS2_science_exposure = LRS2Night[27]
LRS2_science_product = LRS2_science_exposure.quicklook()
LRS2_science_figure = LRS2_science_product.plot(cmap='coolwarm')
```

### Refresh the LRS-2 table later in the night

Run this cell whenever new LRS-2 archives arrive. It updates the existing
night object and displays the refreshed table.


```python
LRS2Night.update()
```
